```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


# 1. Model initialize karein


# 2. Prompt Template banayein
prompt = ChatPromptTemplate.from_template("{topic} ke baare mein ek short line likho.")

# 3. Output Parser banayein
output_parser = StrOutputParser()

# 4. LCEL use karke Chain banayein (Yeh poori chain ek Runnable ban jati hai)
chain = prompt | llm | output_parser

# --- RUNNABLE METHODS KA USE ---

# Method 1: Invoke (Single Input)
print("--- INVOKE EXAMPLE ---")
response = chain.invoke({"topic": "Artificial Intelligence"})
print(response)

# Method 2: Batch (Multiple Inputs)
print("\n--- BATCH EXAMPLE ---")
batch_responses = chain.batch([
    {"topic": "Cricket"},
    {"topic": "Chai"}
])
print(batch_responses)

# Method 3: Stream (Token by Token Output)
print("\n--- STREAM EXAMPLE ---")
for chunk in chain.stream({"topic": "Coding"}):
    print(chunk, end="|", flush=True)
```

# History remembering 

```python
history=[]
while True:
    quary=input("User :")
    if quary.lower() in ["quit","stop","bye"]:
        print("Good bye 🙋‍♂️")
        break

    history.append({"role":"user","content":quary})
    print("User :",quary)

    res=llm.invoke(history)
    history.append({"role":"ai","content":res.content})

    print("AI",res.content)
```
